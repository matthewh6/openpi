```python
pytorch_ckpt_dir = "/gpfs/scrubbed/hongmm/.cache/openpi/openpi-assets/checkpoints/pi0_base_pytorch"
```

```python
from safetensors.torch import load_model as safetensors_load_model
import json
import os

import openpi.models.pi0_config
import openpi.models_pytorch.pi0_pytorch

# Load config
with open(os.path.join(pytorch_ckpt_dir, "config.json")) as f:
    config_dict = json.load(f)

model_config = openpi.models.pi0_config.Pi0Config(
    action_dim=config_dict.get("action_dim"),
    action_horizon=config_dict.get("action_horizon"),
    paligemma_variant=config_dict.get("paligemma_variant"),
    action_expert_variant=config_dict.get("action_expert_variant"),
)

# Instantiate PyTorch model
model = openpi.models_pytorch.pi0_pytorch.PI0Pytorch(model_config)

# Load weights from safetensors file
ckpt_path = os.path.join(pytorch_ckpt_dir, "model.safetensors")
safetensors_load_model(model, ckpt_path)

# Print model architecture
print(model)
```
